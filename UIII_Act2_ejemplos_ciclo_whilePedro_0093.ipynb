#### ciclo while

```python
# Este código suma números ingresados por el usuario hasta que este decide salir ingresando el número 0 ejemplo 1
total = 0                                      # Inicializa la variable para acumular la suma.
numero = 1                                     # Inicializa numero en 1 para poder entrar al ciclo.
while numero != 0:                             # Relacional: el ciclo sigue mientras numero no sea 0.
    numero = int(input("Ingresa un número (0 para salir): ")) # Pide un dato y lo convierte a entero.
    total = total + numero                    # Aritmético: suma el valor ingresado al acumulador total.
print("La suma total es:", total)              # Muestra el resultado final de la suma.
```

```python
# El usuario define hasta qué número desea contar partiendo desde el 1. ejemplo 2
limite = int(input("¿Hasta qué número quieres contar? ")) # El usuario ingresa el tope del conteo.
contador = 1                                              # Establece el punto de partida en 1.
while contador <= limite:                                 # Relacional: se repite mientras no pase el límite.
    print(contador)                                       # Muestra el valor actual del contador.
    contador = contador + 1                               # Aritmético: incrementa el contador de uno en uno.
```

```python
# El ciclo se mantiene activo únicamente si el usuario ingresa números que estén dentro de un rango específico (entre 1 y 10). ejemplo 3
valor = 5                                      # Valor inicial para comenzar el bucle.
while valor >= 1 and valor <= 10:              # Lógico/Relacional: verifica que esté entre 1 y 10.
    print("El número", valor, "está en rango.") # Informa que el valor es correcto.
    valor = int(input("Ingresa otro número entre 1 y 10 para seguir: ")) # Actualiza la condición.
print("Saliste del rango permitido.")          # Mensaje cuando la condición lógica es falsa.
```

```python
# Este ejemplo utiliza un operador lógico para permitir que el usuario decida si quiere repetir un mensaje usando mayúsculas o minúsculas. Ejemplo 4
respuesta = "s"                                # Define la variable con el valor 's' inicial.
while respuesta == "s" or respuesta == "S":    # Lógico/Relacional: acepta 's' minúscula o 'S' mayúscula.
    print("¡Hola como estas!")              # Ejecuta la acción principal del bucle.
    respuesta = input("¿Quieres repetir el mensaje? (s/n): ") # El usuario decide si continúa o no.s
```

```python
# Este ejemplo utiliza un operador lógico para permitir que el usuario decida si quiere repetir un mensaje usando mayúsculas o minúsculas. Ejemplo 4
respuesta = "s"                                # Define la variable con el valor 's' inicial.
while respuesta == "s" or respuesta == "S":    # Lógico/Relacional: acepta 's' minúscula o 'S' mayúscula.
    print("¡Hola como estas!")              # Ejecuta la acción principal del bucle.
    respuesta = input("¿Quieres repetir el mensaje? (s/n): ") # El usuario decide si continúa o no.s
```

```python
# Simula el funcionamiento de una división entera restando el divisor al dividendo hasta que ya no sea posible. Ejemplo 5
dividendo = int(input("Ingresa el dividendo: ")) # Pide el número que será dividido.
divisor = int(input("Ingresa el divisor: "))     # Pide el número que divide.
while dividendo >= divisor:                      # Relacional: resta mientras el dividendo sea mayor o igual.
    dividendo = dividendo - divisor              # Aritmético: resta el divisor sucesivamente.
print("El residuo de la operación es:", dividendo) # El valor sobrante es el residuo.
```
